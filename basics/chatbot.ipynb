```python
# load env 
from dotenv import load_dotenv
load_dotenv()
```

```python
# setup langsmith
from langchain.callbacks.tracers import LangChainTracer

tracer = LangChainTracer()
```

```python
# setup langchain model

import os
from langchain_openai import AzureChatOpenAI

model = AzureChatOpenAI(
    azure_endpoint=os.environ["AZURE_OPENAI_ENDPOINT"],
    azure_deployment=os.environ["AZURE_OPENAI_DEPLOYMENT"],
    openai_api_version=os.environ["AZURE_OPENAI_API_VERSION"],
)
```

```python
# invoke model directly

from langchain_core.messages import HumanMessage

model.invoke([HumanMessage(content="Hi! I'm Bob")])
```

```python
# Model doesn't retain state by default, so if you invoke the model to get the name again.

model.invoke([HumanMessage(content="What's my name ?")], config={"callbacks": [tracer]})
```
